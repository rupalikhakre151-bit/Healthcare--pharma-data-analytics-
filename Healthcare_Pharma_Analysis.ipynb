# Healthcare / Pharmaceutical Data Cleaning & Analysis

This project explores and cleans a pharmaceutical product dataset and performs basic descriptive analysis. It covers data types, missing values, duplicate records, product counts, common dosage forms, routes, marketing categories, and labelers.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/Pharma_data.csv', encoding='latin1', low_memory=False)
df.head()
```

## 1. Dataset Overview


```python
print('Shape:', df.shape)
display(df.info())
display(df.head())
display(df.describe(include='all').T)
```

## 2. Column Types


```python
numeric_columns = df.select_dtypes(include='number').columns.tolist()
categorical_columns = df.select_dtypes(include='object').columns.tolist()
print('Numerical columns:', numeric_columns)
print('Categorical/text columns:', categorical_columns)
```

## 3. Missing Values


```python
missing = df.isnull().sum().sort_values(ascending=False)
display(missing[missing > 0])
```

## 4. Duplicate Records


```python
duplicate_count = df.duplicated().sum()
print('Exact duplicate rows:', duplicate_count)
df = df.drop_duplicates().copy()
print('Shape after duplicate removal:', df.shape)
```

## 5. Basic Pharmaceutical Analysis


```python
print('Unique products:', df['PRODUCTID'].nunique())
print('Unique proprietary names:', df['PROPRIETARYNAME'].nunique())
print('Unique labelers:', df['LABELERNAME'].nunique())

print('\nTop proprietary names:')
display(df['PROPRIETARYNAME'].value_counts().head(10))

print('\nTop dosage forms:')
display(df['DOSAGEFORMNAME'].value_counts().head(10))

print('\nTop routes:')
display(df['ROUTENAME'].value_counts().head(10))

print('\nTop labelers:')
display(df['LABELERNAME'].value_counts().head(10))

print('\nMarketing categories:')
display(df['MARKETINGCATEGORYNAME'].value_counts().head(10))
```

## 6. Visualizations


```python
fig, ax = plt.subplots(figsize=(10,6))
df['DOSAGEFORMNAME'].value_counts().head(10).sort_values().plot(kind='barh', ax=ax)
ax.set_title('Top 10 Dosage Forms')
ax.set_xlabel('Number of Records')
plt.tight_layout()
plt.show()
```
